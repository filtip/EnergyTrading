```python
from pathlib import Path
import pandas as pd
import numpy as np
import importlib
import Ekrt.sf_project.load_data.loader as loader
importlib.reload(loader)

from Ekrt.refactor.prediction_models import optimal_setting
from Ekrt.sf_project.load_data.loader import DataConfig, LoadRawData,PricePanelBuilder, BidAskPanelBuilder, PanelSaver
```

```python
data = optimal_setting.load_project_data() #Load original data

cfg = DataConfig(data_dir=Path("/Users/filipekrt/SF/data")) #path kde jsou úplně původní data
loader = LoadRawData(cfg = cfg)
price_panel = PricePanelBuilder(loader = loader)
bid_ask_panel = BidAskPanelBuilder(loader = loader)
```

```python
#Load prices and bid ask - OUT OF SAMPLE DATA
months_out = ["october","november","december"]
df_bidask_out = bid_ask_panel.combine_months(months = months_out)
dict_prices_out = price_panel.combine_months(months = months_out)

#Save three files
absolute_path = Path("/Users/filipekrt/SF/Ekrt/sf_project/load_data")
PanelSaver.save_price_panel(dict_prices_out, path = absolute_path / "prices_out_of_sample.parquet")
PanelSaver.save_price_long(dict_prices_out, path = absolute_path / "prices_out_of_sample_w_contract_id.parquet")
PanelSaver.save_bidask_panel(df_bidask_out, path = absolute_path / "bidask_out_of_sample.parquet")
```

```python
#Load prices and bid ask - IN SAMPLE
months_in = ["march","april","may","june","july","august","september"]
df_bidask_in = bid_ask_panel.combine_months(months = months_in)
dict_prices_in = price_panel.combine_months(months = months_in)

#Saves three files
PanelSaver.save_price_panel(dict_prices_in, path = absolute_path / "prices_in_sample.parquet")
PanelSaver.save_price_long(dict_prices_in, path = absolute_path / "prices_in_sample_w_contract_id.parquet")
PanelSaver.save_bidask_panel(df_bidask_in, path = absolute_path / "bidask_in_sample.parquet")
```

```python
prices_in = pd.read_parquet(absolute_path / "prices_in_sample.parquet")
prices_out = pd.read_parquet(absolute_path / "prices_out_of_sample.parquet")
prices_w_in = pd.read_parquet(absolute_path / "prices_in_sample_w_contract_id.parquet")
prices_w_out = pd.read_parquet(absolute_path / "prices_out_of_sample_w_contract_id.parquet")
bid_ask_in = pd.read_parquet(absolute_path / "bidask_in_sample.parquet")
bid_ask_out = pd.read_parquet(absolute_path / "bidask_out_of_sample.parquet")


pd.testing.assert_frame_equal(prices_in, data.data_in)
pd.testing.assert_frame_equal(prices_out, data.data_out)
pd.testing.assert_frame_equal(prices_w_in, data.data_w_contract_id_in)
pd.testing.assert_frame_equal(prices_w_out, data.data_w_contract_id_out)
#pd.testing.assert_frame_equal(bid_ask_in, data.bid_ask_in)
#pd.testing.assert_frame_equal(bid_ask_out, data.bid_ask_out)
```
