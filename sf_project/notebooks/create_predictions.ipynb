```python
import settings
import importlib
import create_prediction
importlib.reload(create_prediction)
from create_prediction import Specification, DataTransformation, TrainModel, Prediction, InputsBuilder
import pandas as pd
from pathlib import Path
```

```python
#LOAD DATA
ROOT = Path().resolve().parents[0]
DATA_DIR = ROOT / "sf_project" / "load_data"
data_in = pd.read_parquet(DATA_DIR / "prices_in_sample.parquet")
data_w_contract_id = pd.read_parquet(DATA_DIR / "prices_in_sample_w_contract_id.parquet")
data_out = pd.read_parquet(DATA_DIR / "prices_out_of_sample.parquet")
data_w_contract_id_out = pd.read_parquet(DATA_DIR / "prices_out_of_sample_w_contract_id.parquet")
```

```python
#IN SAMPLE PREDICTIONS
months = ["2025-03", "2025-04", "2025-05", "2025-06", "2025-07", "2025-08", "2025-09"]

for target_col in settings.BEST_PREDICTORS.keys():
    preds = []
    predictor_cols = settings.BEST_PREDICTORS[target_col]["predictor_cols"]

    spec = Specification(target_col=target_col,predictor_cols=predictor_cols)
    data_trans = DataTransformation(spec=spec)
    trainer = TrainModel(spec = spec, data_transformation= data_trans)
    inputs = InputsBuilder(spec = spec)
    predictor = Prediction(trainer = trainer, inputs = inputs)

    for prediction_month in months:
        trainer.fit_on_scenario(data=data_in, prediction_month=prediction_month)
        prediction = predictor.predict_whole_month(data_in, data_w_contract_id, prediction_month)
        preds.append(prediction)

    all_predictions = pd.concat(preds).sort_index()
    #all_predictions.to_parquet(f"/Users/filipekrt/SF/Ekrt/sf_project/predictions_in_sample/{target_col}_predictions.parquet")
```

```python
#OUT OF SAMPLE PREDICTIONS
months = ["2025-10", "2025-11", "2025-12"]

for target_col in settings.BEST_PREDICTORS.keys():
    preds = []
    predictor_cols = settings.BEST_PREDICTORS[target_col]["predictor_cols"]

    spec = Specification(target_col=target_col,predictor_cols=predictor_cols)
    data_trans = DataTransformation(spec=spec)
    trainer = TrainModel(spec = spec, data_transformation= data_trans)
    inputs = InputsBuilder(spec = spec)
    predictor = Prediction(trainer = trainer, inputs = inputs)

    for prediction_month in months:
        trainer.fit_on_scenario_out_of_sample(data_in = data_in, data_out = data_out, prediction_month=prediction_month)
        prediction = predictor.predict_whole_month(data_out, data_w_contract_id_out, prediction_month)
        preds.append(prediction)

    all_predictions = pd.concat(preds).sort_index()
    #all_predictions.to_parquet(f"/Users/filipekrt/SF/Ekrt/sf_project/predictions_out_of_sample/{target_col}_predictions.parquet")
```
